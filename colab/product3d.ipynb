# فيديو منتج ثري دي — شنطة «حمزة»

بيبني الشنطة والأغراض والإضاءة بالكود جوّا Blender، بيعمل ريندر على كرت الشاشة (GPU) تبع Google، وبيطلعلك فيديو 16:9 جاهز بالعناوين والصوت.

**قبل ما تبدأ:** من القائمة فوق اختار **Runtime ← Change runtime type ← T4 GPU ← Save**.

بعدها اضغط ▶ على الخلية اللي تحت (أو **Runtime ← Run all**).

- **preview:** نسخة سريعة بنص الحجم، عشان تشوف الفكرة. حوالي ساعة على T4.
- **full:** النسخة النهائية 1920×1080. ساعات على T4، بس بتكمّل من مكان ما وقفت لو انقطعت الجلسة.
- لو اخترت **حفظ على Drive**، الإطارات والفيديو بينحفظوا بمجلد `product-3d` على Drive، وبتقدر تكمّل بعدين من نفس المكان.

```python
#@title ▶ شغّل الفيديو
QUALITY = "preview"  #@param ["preview", "full"]
SAVE_TO_DRIVE = True  #@param {type:"boolean"}
DESK_PHOTOS = False  #@param {type:"boolean"}
import os, subprocess
if SAVE_TO_DRIVE:
    from google.colab import drive
    drive.mount('/content/drive')
    OUT = '/content/drive/MyDrive/product-3d'
else:
    OUT = '/content/product-3d-out'
REPO = '/content/product-3d-ar'
if not os.path.exists(REPO):
    subprocess.run(['git', 'clone', '-q', '--depth', '1', 'https://github.com/alaahamza465-eng/product-3d-ar', REPO], check=True)
cmd = ['python3', '-u', REPO + '/project/run_all.py', '--quality', QUALITY, '--out', OUT] + (['--stills'] if DESK_PHOTOS else [])
p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
for line in p.stdout: print(line, end='', flush=True)
p.wait()
```

```python
#@title ⬇ نزّل الفيديو على جهازك (بعد ما يخلص)
import glob
from google.colab import files
for f in sorted(glob.glob(OUT + '/product-film-*.mp4')): files.download(f)
```
